Задание 1

```python
def check_car_id(car_id):
    car_id = car_id.upper()
    car_id = car_id.replace('A', 'А').replace('B', 'В').replace('C', 'С')
    car_id = car_id.replace('E', 'Е').replace('H', 'Н').replace('K', 'К')
    car_id = car_id.replace('M', 'М').replace('O', 'О').replace('P', 'Р')
    car_id = car_id.replace('T', 'Т').replace('X', 'Х').replace('Y', 'У')

    result = re.match(r'^([АВЕКМНОРСТУХ]\d{3}[АВЕКМНОРСТУХ]{2})(\d{2,3})$', car_id)

    if result:
        return 'Номер ' + result.group(1) + ' валиден. Регион: ' + result.group(2) + '.'
    else:
        return 'Номер не валиден.'
```

```python
print(check_car_id('А222ВС96'))
print(check_car_id('А123ВЕ777'))
print(check_car_id('АБ22ВВ193'))
print(check_car_id('A222BC96'))
print(check_car_id('А222ВС9'))
print(check_car_id('А222ВС9699'))
print(check_car_id('Я222ВС96'))
print(check_car_id(''))
```

```python
def parse_logs(text):
    result = []

    lines = re.findall(r'^(\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2})\s+(\w+)\s+(.*)$', text, re.M)

    for ts, level, line in lines:
        user = re.search(r'user=(\S+)', line)
        action = re.search(r'action=(\S+)', line)
        ip = re.search(r'ip=(\S+)', line)
        card = re.search(r'card=(\S+)', line)

        user = user.group(1) if user else None
        action = action.group(1) if action else None
        ip = ip.group(1) if ip else None
        card = card.group(1) if card else None
        errors = []

        if not user or not re.match(r'^[\w.+-]+@[\w-]+\.[\w.-]+$', user):
            errors.append('invalid_email')

        if not action:
            errors.append('missing_action')

        ip_pattern = r'^(25[0-5]|2[0-4]\d|1\d\d|[1-9]?\d)(\.(25[0-5]|2[0-4]\d|1\d\d|[1-9]?\d)){3}$'
        if not ip or not re.match(ip_pattern, ip):
            errors.append('invalid_ip')

        if card and re.match(r'^\d{4}-\d{4}-\d{4}-\d{4}$', card):
            card = '****-****-****-' + card[-4:]
        elif card:
            card = None
            errors.append('invalid_card')

        result.append({'ts': ts, 'level': level, 'user': user, 'action': action,
                       'ip': ip, 'card': card, 'errors': errors})

    return result
```

```python
with open('logs_hw4.txt') as file:
    text = file.read()

logs = parse_logs(text)
logs[:3]
```
