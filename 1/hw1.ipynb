```python
import json

purchases = {}

with open('purchase_log.txt', encoding='utf-8') as file:
    next(file)  # пропускаем первую строку с названиями столбцов

    for line in file:
        purchase = json.loads(line)
        purchases[purchase['user_id']] = purchase['category']
```

```python
i = 0

for user_id, category in purchases.items():
    print(user_id, category)
    i += 1

    if i == 2:
       break
```
